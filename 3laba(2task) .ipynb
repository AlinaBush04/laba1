```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard','llll']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']
```

```python
if len(boys) == len(girls):
    boys.sort()
    girls.sort()

    print('Идеальные пары:')
    for boy, girl in zip(boys, girls):
        print(f'{boy} и {girl}')
else:
    print('Внимание, кто-то может остаться без пары!')
```
