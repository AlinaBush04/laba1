```python
import json
import csv

# 1. Создаём словарь с покупками
purchases = {}

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']


# 2. Открываем файл с визитами и создаём новый файл
with open('visit_log.csv', 'r', encoding='utf-8') as visits_file:
    with open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:

        visits = csv.DictReader(visits_file)
        writer = csv.writer(funnel_file)

        # Записываем заголовок
        writer.writerow(['user_id', 'source', 'category'])

        # 3. Обрабатываем визиты построчно
        for visit in visits:
            user_id = visit['user_id']

            if user_id in purchases:
                writer.writerow([
                    user_id,
                    visit['source'],
                    purchases[user_id]
                ])
```
