```python
import pandas as pd
```

```python

# Функция для определения возрастной категории
def check_age(age):
    if age > 25:
        return 'Старше 25'
    else:
        return 'Младше или равен 25'


def load_data(file_name):
    data = pd.read_csv(file_name)
    return data.head(15)


def main():
    data = load_data('web_clients_correct_доп материал к тема 6, 7.csv')

    data['age'] = data['age'].apply(check_age)

    print(data['age'].to_string())


main()
```
