```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
def load_data(file_name):
    data = pd.read_csv(file_name)
    data = data.head(15)

    data['age'] = pd.to_numeric(data['age'], errors='coerce')
    data['bill'] = pd.to_numeric(data['bill'], errors='coerce')

    data = data.dropna(subset=['age', 'bill'])

    return data


# Функция построения диаграммы Seaborn
def create_seaborn_scatter(data):
    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=data,
        x='age',
        y='bill'
    )

    plt.title('Зависимость расходов клиента от возраста')
    plt.xlabel('Возраст клиента')
    plt.ylabel('Расходы клиента')
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# Функция построения диаграммы Plotly
def create_plotly_scatter(data):
    fig = px.scatter(
        data,
        x='age',
        y='bill',
        title='Зависимость расходов клиента от возраста',
        labels={
            'age': 'Возраст клиента',
            'bill': 'Расходы клиента'
        }
    )

    fig.show(renderer='browser')


# Основная функция
def main():
    data = load_data('web_clients_correct_доп материал к тема 6, 7.csv')

    create_seaborn_scatter(data)
    create_plotly_scatter(data)

main()
```
