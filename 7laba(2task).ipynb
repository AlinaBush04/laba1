```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def load_data(file_name):
    data = pd.read_csv(file_name)
    data = data.head(15)

    data['age'] = pd.to_numeric(data['age'], errors='coerce')
    data['bill'] = pd.to_numeric(data['bill'], errors='coerce')

    data = data.dropna(subset=['age', 'bill'])

    return data


# Функция построения линии регрессии
def create_regression_plot(data):
    plt.figure(figsize=(10, 6))

    sns.regplot(
        data=data,
        x='age',
        y='bill'
    )

    plt.title('Регрессия зависимости расходов от возраста')
    plt.xlabel('Возраст клиента')
    plt.ylabel('Расходы клиента')
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# Основная функция
def main():
    data = load_data('web_clients_correct_доп материал к тема 6, 7.csv')

    create_regression_plot(data)


main()
```
