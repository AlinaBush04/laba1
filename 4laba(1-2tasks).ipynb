```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}

```

```python

# 1. Поиск владельца документа
def get_name(documents, number):
    for document in documents:
        if document['number'] == number:
            return document['name']

    return None


# 2. Поиск полки с документом
def get_shelf(directories, number):
    for shelf, numbers in directories.items():
        if number in numbers:
            return shelf

    return None


# 3. Основная функция для обработки команд
def main(documents, directories):
    while True:
        command = input('Введите команду: ')

        if command == 'q':
            print('Программа завершена')
            break

        elif command == 'p':
            number = input('Введите номер документа: ')
            name = get_name(documents, number)

            if name:
                print(f'Владелец документа: {name}')
            else:
                print('Документ не найден')

        elif command == 's':
            number = input('Введите номер документа: ')
            shelf = get_shelf(directories, number)

            if shelf:
                print(f'Документ хранится на полке: {shelf}')
            else:
                print('Документ не найден на полках')

        else:
            print('Неизвестная команда')


main(documents, directories)
```
