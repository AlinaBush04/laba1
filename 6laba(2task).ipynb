```python
import pandas as pd
```

```python
def check_client(row):
    return row['sex'] == 'male' and row['age'] > 30


def load_data(file_name):
    data = pd.read_csv(file_name)
    return data.head(15)


def filter_clients(data):
    result = data[data.apply(check_client, axis=1)]
    return result


def main():
    data = load_data('web_clients_correct_доп материал к тема 6, 7.csv')

    filtered_data = filter_clients(data)

    print(filtered_data.to_string(index=True))


main()
```
